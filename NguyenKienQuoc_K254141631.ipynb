```python
def show_menu():
    print("\nHƯỚNG DẪN CHỨC NĂNG")
    print("=" * 60)
    print("1. Thêm khách hàng")
    print("2. Xóa khách hàng")
    print("3. Hiển thị bảng xếp hạng")
    print("4. Thoát chương trình")
    print("=" * 60)
def add_customer(customers):
    customer_data = get_customer_data(len(customers) + 1)
    if customer_data:
        score = calculate_credit_score(
            customer_data['income_level'],
            customer_data['saving_rate'],
            customer_data['payment_history']
        )
        rank = classify_score(score)
        customers.append({
            'name': customer_data['name'],
            'score': score,
            'rank': rank
        })
        print(f"Đã thêm khách hàng: {customer_data['name']}")
def remove_customer(customers):
    if not customers:
        print("Không có khách hàng nào để xóa.")
        return
    print("\nDanh sách khách hàng hiện tại:")
    for idx, c in enumerate(customers, 1):
        print(f"{idx}. {c['name']} ({c['score']:.2f} điểm, loại {c['rank']})")
    try:
        choice = int(input("Nhập số thứ tự khách hàng muốn xóa: "))
        if 1 <= choice <= len(customers):
            removed = customers.pop(choice - 1)
            print(f"Đã xóa khách hàng: {removed['name']}")
        else:
            print("Số thứ tự không hợp lệ.")
    except ValueError:
        print("Vui lòng nhập số nguyên hợp lệ.")
def classify_score(score):
    if not (0 <= score <= 100):
        raise ValueError(f"Score {score} is out of valid range (0-100)")
    if score >= 80:
        return 'A'
    elif score >= 65:
        return 'B'
    elif score >= 50:
        return 'C'
    else:
        return 'D'
def get_valid_input(prompt, min_val=0, max_val=100):
    while True:
        try:
            # User data nhận dấu tiếng việt
            user_input = input(prompt).strip()

            if user_input.lower() in ['exit', 'quit', 'thoat']:
                return None
            value = float(user_input)
            if min_val <= value <= max_val:
                return value
            else:
                print(f"Giá trị phải nằm trong khoảng {min_val} đến {max_val}.")
        except ValueError:
            print("Vui lòng nhập một số hợp lệ.")
        except KeyboardInterrupt:
            print("\nĐã hủy nhập liệu.")
            raise
def calculate_credit_score(income_level, saving_rate, payment_history):
    INCOME_WEIGHT = 0.4
    SAVING_WEIGHT = 0.3
    PAYMENT_WEIGHT = 0.3
    total_weight = INCOME_WEIGHT + SAVING_WEIGHT + PAYMENT_WEIGHT
    if abs(total_weight - 1.0) > 0.0001:
        raise ValueError(f"Weights must sum to 1.0, got {total_weight}")
    score = (INCOME_WEIGHT * income_level +
             SAVING_WEIGHT * saving_rate +
             PAYMENT_WEIGHT * payment_history)
    return round(score, 2)
def get_customer_count():
    while True:
        try:
            user_input = input("Nhập số lượng khách hàng (hoặc 'exit' để thoát): ").strip()
            if user_input.lower() in ['exit', 'quit', 'thoat']:
                return None
            n = int(user_input)
            if n > 0:
                return n
            else:
                print("Số lượng khách hàng phải lớn hơn 0.")
        except ValueError:
            print("Vui lòng nhập một số nguyên hợp lệ.")
        except KeyboardInterrupt:
            print("\nĐã hủy nhập liệu.")
            raise
def get_customer_data(customer_number):
    print(f"\n--- Khách hàng {customer_number} ---")
    while True:
        name = input("  Họ tên (hoặc 'exit' để thoát): ").strip()
        if name.lower() in ['exit', 'quit', 'thoat']:
            return None
        if name:
            break
        print("  Họ tên không được để trống.")
    income_level = get_valid_input("  Thu nhập (0-100) (income_level) [exit để thoát]: ")
    if income_level is None:
        return None
    saving_rate = get_valid_input("  Tỷ lệ tiết kiệm (0-100) (saving_rate) [exit để thoát]: ")
    if saving_rate is None:
        return None
    payment_history = get_valid_input("  Lịch sử thanh toán (0-100) (payment_history) [exit để thoát]: ")
    if payment_history is None:
        return None
    return {
        'name': name,
        'income_level': income_level,
        'saving_rate': saving_rate,
        'payment_history': payment_history
    }
def display_results(customers):
    print("\n" + "═" * 67)
    print("║{:^64} ║".format("BẢNG XẾP HẠNG TÍN DỤNG KHÁCH HÀNG"))
    print("╠═════╦═════════════════════════════════╦════════════╦════════════╣")
    print("║{:^5}║{:^33}║{:^12}║{:^12}║".format("STT", "Họ tên", "Điểm", "Xếp loại"))
    print("╠═════╬═════════════════════════════════╬════════════╬════════════╣")
    for idx, c in enumerate(customers, 1):
        print("║{:>5}║ {:<32}║{:>12.2f}║{:>12}║".format(
            idx, c['name'][:32], c['score'], c['rank']
        ))
    print("╚═════╩═════════════════════════════════╩════════════╩════════════╝")
    if customers:
        avg_score = sum(c['score'] for c in customers) / len(customers)
        print(f"\nĐiểm trung bình: {avg_score:.2f}")
        max_score = max(c['score'] for c in customers)
        min_score = min(c['score'] for c in customers)
        top_customers = [c for c in customers if c['score'] == max_score]
        low_customers = [c for c in customers if c['score'] == min_score]
        print(f"Điểm cao nhất : {max_score:.2f} -", end=" ")
        print(", ".join(c['name'] for c in top_customers))
        print(f"Điểm thấp nhất: {min_score:.2f} -", end=" ")
        print(", ".join(c['name'] for c in low_customers))
def main():
    customers = []
    try:
        print("=" * 60)
        print("       HỆ THỐNG XẾP HẠNG TÍN DỤNG KHÁCH HÀNG")
        print("=" * 60)
        print("Gõ 'exit', 'quit', hoặc 'thoat' bất kỳ lúc nào để thoát chương trình\n")

        n = get_customer_count()
        if n is None:
            print("\nĐã thoát chương trình.")
            return
        for i in range(n):
            try:
                customer_data = get_customer_data(i + 1)
                if customer_data is None:
                    print("\nĐã thoát quá trình nhập liệu.")
                    break
                score = calculate_credit_score(
                    customer_data['income_level'],
                    customer_data['saving_rate'],
                    customer_data['payment_history']
                )
                rank = classify_score(score)
                customers.append({
                    'name': customer_data['name'],
                    'score': score,
                    'rank': rank
                })
            except Exception as e:
                print(f"Lỗi khi xử lý khách hàng {i + 1}: {e}")
                choice = input("Bỏ qua khách hàng này? (y/n hoặc 'exit' để thoát): ").strip().lower()
                if choice in ['exit', 'quit', 'thoat']:
                    break
                elif choice != 'n':
                    continue
        while True:
            show_menu()
            choice = input("Nhập lựa chọn của bạn (1-4): ").strip()
            if choice == '1':
                add_customer(customers)
            elif choice == '2':
                remove_customer(customers)
            elif choice == '3':
                if customers:
                    customers.sort(key=lambda x: x['score'], reverse=True)
                    display_results(customers)
                else:
                    print("Chưa có dữ liệu khách hàng.")
            elif choice == '4':
                print("Đã thoát chương trình.")
                break
            else:
                print("Lựa chọn không hợp lệ. Vui lòng chọn từ 1 đến 4.")
    except KeyboardInterrupt:
        print("\n\nChương trình đã bị hủy bởi người dùng.")
    except Exception as e:
        print(f"\nLỗi không mong đợi: {e}")
if __name__ == "__main__":
    while True:
        try:
            main()
            cont = input("\nBạn có muốn chạy lại chương trình không? (y/n): ").strip().lower()
            if cont == 'n':
                print("Cảm ơn bạn đã sử dụng chương trình. Tạm biệt!")
                break
            elif cont != 'y':
                print("Vui lòng nhập 'y' để tiếp tục hoặc 'n' để thoát.")
        except KeyboardInterrupt:
            print("\nChương trình đã bị hủy bởi người dùng.")
        except Exception as e:
            print(f"\nLỗi không mong đợi: {e}")
```

### **Bài 2**

```python
import math
import datetime

def nhap_lieu_hop_le(prompt, min_val=float('-inf'), max_val=float('inf')):
    while True:
        try:
            user_input = input(prompt).strip()
            if user_input.lower() in ['exit', 'quit', 'thoat']:
                return None
            value = float(user_input)

            if min_val <= value <= max_val:
                return value
            else:
                print(f"Giá trị phải nằm trong khoảng {min_val} đến {max_val}.")
        except ValueError:
            print("Vui lòng nhập một số hợp lệ.")
        except KeyboardInterrupt:
            print("\nĐã hủy nhập liệu.")
            raise
def nhap_ten_co_phieu(prompt):
    while True:
        user_input = input(prompt).strip()
        if user_input.lower() in ['exit', 'quit', 'thoat']:
            return None
        if user_input:
            return user_input.upper()
        print("Mã cổ phiếu không được để trống.")
def hoi_luu_file(prompt):
    while True:
        choice = input(prompt).strip().lower()
        if choice in ['y', 'yes', 'co']:
            return True
        if choice in ['n', 'no', 'khong', 'k']:
            return False
        if choice in ['exit', 'quit', 'thoat']:
            return False  # Coi như không lưu
        print("Vui lòng nhập 'y' (đồng ý) hoặc 'n' (không).")
def luu_ket_qua_ra_file(danh_muc, loi_suat_ky_vong, do_lech_chuan_danh_muc, filename="lich_su_phan_tich.txt"):
    now = datetime.datetime.now()
    timestamp = now.strftime("%Y-%m-%d %H:%M:%S")
    canh_bao = ""
    if do_lech_chuan_danh_muc > 10:
        canh_bao = "⚠️  Danh mục rủi ro cao"
    else:
        canh_bao = "✅  Danh mục ổn định"
    try:
        with open(filename, 'a', encoding='utf-8') as f:
            f.write(f"========= PHÂN TÍCH LÚC: {timestamp} =========\n")
            f.write("I. DANH MỤC CỔ PHIẾU:\n")
            f.write(f"{'Mã CP':<10} | {'Tỷ trọng (%)':>12} | {'Lợi suất (%)':>12} | {'Độ lệch CV (%)':>15}\n")
            f.write("-" * 54 + "\n")
            tong_ty_trong = 0.0
            for cp in danh_muc:
                ty_trong_percent = cp['ty_trong'] * 100
                f.write(
                    f"{cp['ma']:<10} | {ty_trong_percent:>12.2f} | {cp['loi_suat']:>12.2f} | {cp['do_lech_chuan']:>15.2f}\n")
                tong_ty_trong += cp['ty_trong']
            f.write("-" * 54 + "\n")
            f.write(f"Tổng tỷ trọng: {tong_ty_trong * 100:.2f}%\n\n")
            f.write("II. KẾT QUẢ PHÂN TÍCH:\n")
            f.write(f"  Lợi suất kỳ vọng (E(Rp)): {loi_suat_ky_vong:.2f}%\n")
            f.write(f"  Độ lệch chuẩn (Risk, σp): {do_lech_chuan_danh_muc:.2f}%\n")
            f.write(f"  Cảnh báo: {canh_bao}\n")
            f.write("=" * (57 + len(timestamp)) + "\n\n")  # Dấu phân cách
        print(f"\n✅  Đã lưu kết quả phân tích vào file '{filename}'")
    except Exception as e:
        print(f"\n LỖI: Không thể lưu file. Lý do: {e}")
def nhap_du_lieu_co_phieu():
    print("\n--- Nhập thông tin cổ phiếu (gõ 'exit' để hủy) ---")
    ma = nhap_ten_co_phieu("  Mã cổ phiếu: ")
    if ma is None: return None
    ty_trong_percent = nhap_lieu_hop_le("  Tỷ trọng (%): ", 0, 100)
    if ty_trong_percent is None: return None
    loi_suat = nhap_lieu_hop_le("  Lợi suất kỳ vọng (%): ", max_val=1000)
    if loi_suat is None: return None
    do_lech_chuan = nhap_lieu_hop_le("  Độ lệch chuẩn (%): ", 0, 1000)
    if do_lech_chuan is None: return None
    return {
        'ma': ma,
        'ty_trong': ty_trong_percent / 100.0,
        'loi_suat': loi_suat,
        'do_lech_chuan': do_lech_chuan
    }
def hien_thi_danh_muc(danh_muc):
    if not danh_muc:
        print("\nDanh mục hiện đang trống.")
        return
    print("\n" + "═" * 66)
    print("║{:^64}║".format("DANH MỤC CỔ PHIẾU HIỆN TẠI"))
    print("╠═════╦═════════════╦═════════════╦══════════════╦═══════════════╣")
    print("║{:^5}║{:^13}║{:^13}║{:^14}║{:^15}║".format(
        "STT", "Mã CP", "Tỷ trọng (%)", "Lợi suất (%)", "Độ lệch CV (%)"
    ))
    print("╠═════╬═════════════╬═════════════╬══════════════╬═══════════════╣")
    tong_ty_trong = 0.0
    for idx, cp in enumerate(danh_muc, 1):
        ty_trong_percent = cp['ty_trong'] * 100
        print("║{:>5}║ {:<12}║{:>12.2f} ║{:>13.2f} ║{:>14.2f} ║".format(
            idx, cp['ma'], ty_trong_percent, cp['loi_suat'], cp['do_lech_chuan']
        ))
        tong_ty_trong += cp['ty_trong']
    print("╚═════╩═════════════╩═════════════╩══════════════╩═══════════════╝")
    print(f"\nTổng tỷ trọng hiện tại: {tong_ty_trong * 100:.2f}%")
    if not math.isclose(tong_ty_trong, 1.0):
        print("⚠️⚠️⚠️ CẢNH BÁO: Tổng tỷ trọng không bằng 100%. Kết quả phân tích sẽ không chính xác.")
    else:
        print("✅Tổng tỷ trọng đã đủ 100%.")
def phan_tich_danh_muc(danh_muc):
    if not danh_muc:
        print("Danh mục trống, không thể phân tích.")
        return
    print("\n--- Đang tiến hành phân tích ---")
    tong_ty_trong = sum(cp['ty_trong'] for cp in danh_muc)
    if not math.isclose(tong_ty_trong, 1.0, abs_tol=0.001):
        print(f"LỖI: Tổng tỷ trọng là {tong_ty_trong * 100:.2f}%, không phải 100%.")
        print("Vui lòng điều chỉnh danh mục trước khi phân tích.")
        return
    loi_suat_ky_vong = sum(cp['ty_trong'] * cp['loi_suat'] for cp in danh_muc)

    do_lech_chuan_danh_muc = math.sqrt(
        sum((cp['ty_trong'] ** 2) * (cp['do_lech_chuan'] ** 2) for cp in danh_muc)
    )
    print("\n--- Kết quả phân tích danh mục ---")
    print(f"  Lợi suất kỳ vọng (E(Rp)): {loi_suat_ky_vong:.2f}%")
    print(f"  Độ lệch chuẩn (Risk, σp): {do_lech_chuan_danh_muc:.2f}%")
    if do_lech_chuan_danh_muc > 10:
        print("  Cảnh báo: ⚠️Danh mục rủi ro cao")
    else:
        print("  Cảnh báo: ✅Danh mục ổn định")
    print("-" * 30)
    try:
        if hoi_luu_file("Bạn có muốn lưu kết quả này vào file nhật ký? (y/n): "):
            luu_ket_qua_ra_file(
                danh_muc,
                loi_suat_ky_vong,
                do_lech_chuan_danh_muc
            )
    except Exception as e:
        print(f"Đã xảy ra lỗi khi lưu file: {e}")
def show_menu():
    print("\n" + "=" * 30)
    print("  MENU QUẢN LÝ PORTFOLIO")
    print("=" * 30)
    print("1. Thêm cổ phiếu vào danh mục")
    print("2. Xóa cổ phiếu khỏi danh mục")
    print("3. Hiển thị danh mục hiện tại")
    print("4. Phân tích rủi ro danh mục")
    print("5. Thoát chương trình")
    print("-" * 30)
def them_co_phieu(danh_muc):
    try:
        co_phieu_moi = nhap_du_lieu_co_phieu()
        if co_phieu_moi:
            danh_muc.append(co_phieu_moi)
            print(f"Đã thêm thành công cổ phiếu: {co_phieu_moi['ma']}")
            hien_thi_danh_muc(danh_muc)
        else:
            print("Đã hủy thao tác thêm cổ phiếu.")
    except KeyboardInterrupt:
        print("\nĐã hủy thao tác thêm cổ phiếu.")
def xoa_co_phieu(danh_muc):
    if not danh_muc:
        print("Danh mục trống, không có gì để xóa.")
        return
    print("\n--- Xóa cổ phiếu ---")
    hien_thi_danh_muc(danh_muc)
    try:
        prompt = f"Nhập STT cổ phiếu muốn xóa (1-{len(danh_muc)}) (hoặc 'exit' để hủy): "
        lua_chon_stt = nhap_lieu_hop_le(prompt, 1, len(danh_muc))
        if lua_chon_stt is None:
            print("Đã hủy thao tác xóa.")
            return
        index_can_xoa = int(lua_chon_stt) - 1
        removed = danh_muc.pop(index_can_xoa)
        print(f"Đã xóa thành công cổ phiếu: {removed['ma']}")
        hien_thi_danh_muc(danh_muc)
    except (ValueError, IndexError):
        print("STT không hợp lệ.")
    except KeyboardInterrupt:
        print("\nĐã hủy thao tác xóa.")
def main():
    danh_muc = []
    print("=" * 60)
    print("       CHƯƠNG TRÌNH PHÂN TÍCH RỦI RO PORTFOLIO")
    print("=" * 60)
    print("Gõ 'exit', 'quit', hoặc 'thoat' bất kỳ lúc nào để hủy thao tác.")
    try:
        while True:
            show_menu()
            choice = input("Nhập lựa chọn của bạn (1-5): ").strip()
            if choice == '1':
                them_co_phieu(danh_muc)
            elif choice == '2':
                xoa_co_phieu(danh_muc)
            elif choice == '3':
                hien_thi_danh_muc(danh_muc)
            elif choice == '4':
                phan_tich_danh_muc(danh_muc)
            elif choice == '5':
                print("\nĐã thoát chương trình quản lý portfolio.")
                break
            else:
                print("Lựa chọn không hợp lệ. Vui lòng chọn từ 1 đến 5.")
    except KeyboardInterrupt:
        print("\n\nChương trình đã bị hủy bởi người dùng.")
    except Exception as e:
        print(f"\nLỗi không mong đợi: {e}")
if __name__ == "__main__":
    while True:
        try:
            main()
            cont = input("\nBạn có muốn chạy lại toàn bộ chương trình không? (y/n): ").strip().lower()
            if cont == 'n':
                print("Cảm ơn bạn đã sử dụng chương trình. Tạm biệt!")
                break
            elif cont != 'y':
                print("Vui lòng nhập 'y' để tiếp tục hoặc 'n' để thoát.")
        except KeyboardInterrupt:
            print("\n\nChương trình đã bị hủy bởi người dùng.")
            break
        except Exception as e:
            print(f"\nLỗi nghiêm trọng: {e}")
            break
```

```python
import datetime
import getpass
import pickle
import os
import re
class LuiMotBuocException(Exception):
    pass
def clear_screen():
    os.system('cls' if os.name == 'nt' else 'clear')
def nhap_lieu_hop_le(prompt, min_val=float('-inf'), max_val=float('inf'), la_so_nguyen=False):
    prompt_full = f"{prompt} (gõ 'lui' để quay lại, 'exit' để hủy): "
    while True:
        try:
            user_input = input(prompt_full).strip()
            if user_input.lower() in ['exit', 'quit', 'thoat']:
                return None
            if user_input.lower() in ['lui', 'back', 'esp']:
                raise LuiMotBuocException()
            if la_so_nguyen:
                value = int(user_input)
            else:
                value = float(user_input)
            if min_val <= value <= max_val:
                return value
            else:
                print(f"Giá trị phải nằm trong khoảng {min_val} đến {max_val}.")
        except ValueError:
            if la_so_nguyen:
                print("Vui lòng nhập một số NGUYÊN hợp lệ.")
            else:
                print("Vui lòng nhập một SỐ hợp lệ.")
        except KeyboardInterrupt:
            print("\nĐã hủy nhập liệu.")
            raise
def nhap_van_ban(prompt, yeu_cau_mat_khau=False, kiem_tra_ten=False):
    prompt_full = f"{prompt} (gõ 'lui' để quay lại, 'exit' để hủy): "
    while True:
        try:
            user_input = ""
            if yeu_cau_mat_khau:
                user_input = getpass.getpass(prompt_full).strip()
            else:
                user_input = input(prompt_full).strip()
            if user_input.lower() in ['exit', 'quit', 'thoat']:
                return None
            if user_input.lower() in ['lui', 'back', 'esp']:
                raise LuiMotBuocException()
            if not user_input:
                print("Nội dung không được để trống.")
                continue
            return user_input
        except KeyboardInterrupt:
            print("\nĐã hủy nhập liệu.")
            raise
class TaiKhoan:
    PHI_RUT_TIEN_PERCENT = 0.005
    PHI_CHUYEN_TIEN_PERCENT = 0.005
    def __init__(self, so_tai_khoan, ten_chu_tai_khoan, mat_khau, so_du_ban_dau, cau_hoi, tra_loi):
        self.so_tai_khoan = so_tai_khoan
        self.ten_chu_tai_khoan = ten_chu_tai_khoan
        self.mat_khau = mat_khau
        self.cau_hoi_bao_mat = cau_hoi
        self.tra_loi_bao_mat = tra_loi
        if so_du_ban_dau < 0:
            self.so_du = 0
            self.lich_su = [f"Tạo tài khoản thất bại (số dư âm). Đặt về 0."]
        else:
            self.so_du = so_du_ban_dau
            self.lich_su = [f"[{self._thoi_gian()}] Tạo tài khoản. Số dư: {self.so_du:,.2f} VND"]
    def _thoi_gian(self):
        return datetime.datetime.now().strftime("%Y-%m-%d %H:%M")
    def kiem_tra_mat_khau(self, mat_khau_nhap_vao):
        return self.mat_khau == mat_khau_nhap_vao
    def kiem_tra_cau_tra_loi(self, tra_loi_nhap_vao):
        return self.tra_loi_bao_mat.lower() == tra_loi_nhap_vao.lower()
    def nap_tien(self, so_tien):
        if so_tien <= 0:
            return (False, f"[LỖI] Số tiền nạp phải lớn hơn 0.")
        self.so_du += so_tien
        thong_bao = f"[{self._thoi_gian()}] Nạp {so_tien:,.2f} VND. Số dư mới: {self.so_du:,.2f} VND"
        self.lich_su.append(thong_bao)
        return (True, f"[OK] {thong_bao}")
    def rut_tien(self, so_tien):
        if so_tien <= 0:
            return (False, f"[LỖI] Số tiền rút phải lớn hơn 0.")
        phi = so_tien * self.PHI_RUT_TIEN_PERCENT
        tong_can_tru = so_tien + phi
        if self.so_du >= tong_can_tru:
            self.so_du -= tong_can_tru
            thong_bao = f"[{self._thoi_gian()}] Rút {so_tien:,.2f} VND (Phí {phi:,.2f} VND). Số dư mới: {self.so_du:,.2f} VND"
            self.lich_su.append(thong_bao)
            return (True, f"[OK] {thong_bao}")
        else:
            thong_bao = f"[{self._thoi_gian()}] Rút {so_tien:,.2f} VND thất bại (Không đủ số dư. Cần {tong_can_tru:,.2f} VND)."
            self.lich_su.append(thong_bao)
            return (False, f"[LỖI] {thong_bao}")
    def chuyen_tien(self, den_tai_khoan, so_tien):
        if so_tien <= 0:
            return (False, f"[LỖI] Số tiền chuyển phải lớn hơn 0.")
        if self == den_tai_khoan:
            return (False, f"[LỖI] Không thể tự chuyển khoản cho chính mình.")
        phi = so_tien * self.PHI_CHUYEN_TIEN_PERCENT
        tong_can_tru = so_tien + phi
        if self.so_du >= tong_can_tru:
            self.so_du -= tong_can_tru
            den_tai_khoan.so_du += so_tien
            thong_bao_gui = f"[{self._thoi_gian()}] Chuyển {so_tien:,.2f} VND đến STK {den_tai_khoan.so_tai_khoan} ({den_tai_khoan.ten_chu_tai_khoan}). Phí {phi:,.2f} VND. Số dư mới: {self.so_du:,.2f} VND"
            thong_bao_nhan = f"[{self._thoi_gian()}] Nhận {so_tien:,.2f} VND từ STK {self.so_tai_khoan} ({self.ten_chu_tai_khoan}). Số dư mới: {den_tai_khoan.so_du:,.2f} VND"
            self.lich_su.append(thong_bao_gui)
            den_tai_khoan.lich_su.append(thong_bao_nhan)
            return (True, f"[OK] {thong_bao_gui}")
        else:
            thong_bao = f"[{self._thoi_gian()}] Chuyển {so_tien:,.2f} VND đến STK {den_tai_khoan.so_tai_khoan} thất bại (Không đủ số dư. Cần {tong_can_tru:,.2f} VND)."
            self.lich_su.append(thong_bao)
            return (False, f"[LỖI] {thong_bao}")
    def in_sao_ke(self):
        clear_screen()
        print("\n" + "═" * 80)
        print(f"║ {'SAO KÊ TÀI KHOẢN':^76} ║")
        print(f"║ STK: {self.so_tai_khoan:<72}║")
        print(f"║ Tên: {self.ten_chu_tai_khoan:<72}║")
        print(f"║ Số dư hiện tại: {self.so_du:,.2f} VND {'':<47}║")
        print(f"║ Tổng số giao dịch: {len(self.lich_su):<58}║")
        print("╠" + "═" * 78 + "╣")
        print(f"║ {'LỊCH SỬ GIAO DỊCH':^76} ║")
        print("╠" + "═" * 78 + "╣")
        if not self.lich_su:
            print("║ {'Không có lịch sử giao dịch':^76} ║")
        else:
            for dong in self.lich_su:
                print(f"║ {dong:<76.76} ║")
        print("╚" + "═" * 78 + "╝")
        input("\nNhấn Enter để quay lại menu...")
    def luu_sao_ke_ra_file(self):
        filename = f"sao_ke_{self.so_tai_khoan}_{datetime.datetime.now().strftime('%Y%m%d_%H%M%S')}.txt"
        try:
            with open(filename, 'w', encoding='utf-8') as f:
                f.write("=" * 80 + "\n")
                f.write(f"SAO KÊ TÀI KHOẢN\n")
                f.write(f"Ngày xuất: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
                f.write("-" * 80 + "\n")
                f.write(f"STK: {self.so_tai_khoan}\n")
                f.write(f"Tên chủ tài khoản: {self.ten_chu_tai_khoan}\n")
                f.write(f"Số dư cuối kỳ: {self.so_du:,.2f} VND\n")
                f.write(f"Tổng số giao dịch: {len(self.lich_su)}\n")
                f.write("=" * 80 + "\n")
                f.write("LỊCH SỬ GIAO DỊCH CHI TIẾT:\n")
                f.write("-" * 80 + "\n")
                if not self.lich_su:
                    f.write("Không có lịch sử giao dịch.\n")
                else:
                    for dong in self.lich_su:
                        f.write(f"- {dong}\n")
                f.write("\n" + "=" * 80 + "\n")
                f.write("Cảm ơn quý khách đã sử dụng dịch vụ.\n")
            print(f"\n✅  Đã xuất sao kê thành công ra file: '{filename}'")
        except Exception as e:
            print(f"\n⚠️  LỖI: Không thể lưu file sao kê. Lý do: {e}")
class NganHang:
    def __init__(self):
        self.danh_sach_tai_khoan = {}
        self.next_account_id = 100001
    def tao_so_tai_khoan_moi(self):
        while str(self.next_account_id) in self.danh_sach_tai_khoan:
            self.next_account_id += 1
        return str(self.next_account_id)
    def them_tai_khoan(self, ten, mat_khau, so_du_ban_dau, cau_hoi, tra_loi):
        stk = self.tao_so_tai_khoan_moi()
        if so_du_ban_dau < 0:
            print("[LỖI] Số dư ban đầu không thể âm. Đã đặt về 0.")
            so_du_ban_dau = 0
        tai_khoan_moi = TaiKhoan(stk, ten, mat_khau, so_du_ban_dau, cau_hoi, tra_loi)
        self.danh_sach_tai_khoan[stk] = tai_khoan_moi
        return tai_khoan_moi
    def tim_tai_khoan(self, so_tai_khoan):
        return self.danh_sach_tai_khoan.get(so_tai_khoan)
    def dang_nhap(self, so_tai_khoan, mat_khau):
        tk = self.tim_tai_khoan(so_tai_khoan)
        if tk and tk.kiem_tra_mat_khau(mat_khau):
            return tk
        return None
    def hien_thi_danh_sach(self):
        clear_screen()
        print("\n" + "═" * 67)
        print("║{:^65}║".format("BẢNG KÊ TÀI KHOẢN QUOCBANK"))
        print("╠══════════╦══════════════════════════════╦═══════════════════════╣")
        print("║{:^10}║{:^30}║{:^20}   ║".format("STK", "Chủ tài khoản", "Số dư (VND)"))
        print("╠══════════╬══════════════════════════════╬═══════════════════════╣")
        tong_so_du = 0.0
        so_luong_tk = 0
        if not self.danh_sach_tai_khoan:
            print("║{:^64} ║".format("Ngân hàng chưa có tài khoản nào"))
        else:
            for stk, tk in self.danh_sach_tai_khoan.items():
                print("║ {:<9}║ {:<29}║{:>19,.2f}    ║".format(
                    stk, tk.ten_chu_tai_khoan[:29], tk.so_du
                ))
                tong_so_du += tk.so_du
                so_luong_tk += 1
        print("╠══════════╩══════════════════════════════╩═══════════════════════╣")
        print("║ {:<41} {:>21} ║".format("Tổng số tài khoản:", so_luong_tk))
        print("║ {:<41} {:>21,.2f} ║".format("Tổng số dư toàn ngân hàng (VND):", tong_so_du))
        print("╚" + "═" * 64 + "═╝")
        input("\nNhấn Enter để quay lại menu...")
    def hien_thi_thong_tin_dang_nhap(self):
        clear_screen()
        print("\n" + "!" * 88)
        print("         ẽ!" + " CẢNH BÁO: BẠN ĐANG XEM THÔNG TIN BẢO MẬT CỦA TOÀN BỘ NGÂN HÀNG " + "!")
        print("!" * 88)
        print("\n" + "═" * 90)
        print("║{:^88}║".format("BẢNG KÊ THÔNG TIN ĐĂNG NHẬP VÀ BẢO MẬT"))
        print("╠══════════╦═══════════════════╦═══════════════╦══════════════════════╦══════════════════╣")
        print("║{:^10}║{:^19}║{:^15}║{:^22}║{:^18}║".format(
            "STK", "Chủ tài khoản", "Mật khẩu", "Câu hỏi bảo mật", "Câu trả lời"
        ))
        print("╠══════════╬═══════════════════╬═══════════════╬══════════════════════╬══════════════════╣")
        if not self.danh_sach_tai_khoan:
            print("║{:^88}║".format("Ngân hàng chưa có tài khoản nào"))
        else:
            for stk, tk in self.danh_sach_tai_khoan.items():
                print("║ {:<9}║ {:<18}║ {:<14}║ {:<21}║ {:<17}║".format(
                    stk,
                    tk.ten_chu_tai_khoan[:18],
                    tk.mat_khau[:14],
                    tk.cau_hoi_bao_mat[:21],
                    tk.tra_loi_bao_mat[:17]
                ))
        print("╚══════════╩═══════════════════╩═══════════════╩══════════════════════╩══════════════════╝")
        input("\nNhấn Enter để quay lại menu...")
def hien_thi_tro_giup():
    clear_screen()
    print("\n" + "═" * 80)
    print("║{:^78}║".format("HỖ TRỢ VÀ GIẢI ĐÁP LỖI THƯỜNG GẶP - QUOCBANK"))
    print("╠" + "═" * 78 + "╣")
    print("║ 1. Lỗi: 'STK hoặc Mật khẩu không chính xác' khi đăng nhập.                   ║")
    print("║    -> Cách khắc phục:                                                        ║")
    print("║       - Kiểm tra kỹ STK (Số Tài Khoản) bạn đã nhập.                          ║")
    print("║       - Nếu chắc chắn STK đúng, hãy dùng chức năng '3. Quên mật khẩu'.       ║")
    print("║       - Nếu không nhớ STK, liên hệ Admin (chức năng '4') để xem lại.         ║")
    print("╠" + "─" * 78 + "╣")
    print("║ 2. Lỗi: 'STK [số] không tồn tại' khi chuyển tiền.                            ║")
    print("║    -> Cách khắc phục:                                                        ║")
    print("║       - Yêu cầu người nhận xác nhận lại STK của họ.                          ║")
    print("║       - Đảm bảo bạn không nhập nhầm (STK của bank này bắt đầu từ 100001).    ║")
    print("╠" + "─" * 78 + "╣")
    print("║ 3. Lỗi: 'Không đủ số dư' khi rút/chuyển tiền.                                ║")
    print("║    -> Cách khắc phục:                                                        ║")
    print("║       - Bạn cần chừa lại 0.5% (Phí) trên tổng số tiền rút/chuyển.            ║")
    print("║       - Ví dụ: Muốn rút 1,000,000 VND, bạn cần có ít nhất 1,005,000 VND.     ║")
    print("║       - Hãy nạp thêm tiền hoặc rút/chuyển số tiền nhỏ hơn.                   ║")
    print("╠" + "─" * 78 + "╣")
    print("║ 4. Lỗi: 'Câu trả lời không chính xác' khi dùng 'Quên mật khẩu'.              ║")
    print("║    -> Cách khắc phục:                                                        ║")
    print("║       - Hệ thống không phân biệt chữ HOA/thường, nhưng hãy kiểm tra          ║")
    print("║         chính tả, dấu và khoảng trắng.                                       ║")
    print("║       - Nếu quên, không thể khôi phục. Liên hệ Admin để xóa TK và tạo lại.   ║")
    print("╠" + "─" * 78 + "╣")
    print("║ 5. Cách Hủy / Quay lại (Lùi 1 bước):                                         ║")
    print("║    -> Cách khắc phục:                                                        ║")
    print("║       - Gõ 'lui' (hoặc 'back', 'esp') để quay lại bước nhập liệu *trước đó*. ║")
    print("║       - Gõ 'exit' (hoặc 'quit', 'thoat') để *hủy bỏ* thao tác và             ║")
    print("║         quay về menu chính.                                                  ║")
    print("╚" + "═" * 78 + "╝")
    input("\nNhấn Enter để quay lại menu...")
def luu_du_lieu_ngan_hang(ngan_hang, filename="ngan_hang_data.pkl"):
    try:
        with open(filename, 'wb') as f:
            pickle.dump(ngan_hang, f)
        print(f"\n[Dữ liệu ngân hàng đã được lưu vào '{filename}']")
    except Exception as e:
        print(f"\n[LỖI] Không thể lưu dữ liệu ngân hàng: {e}")
def tai_du_lieu_ngan_hang(filename="ngan_hang_data.pkl"):
    if os.path.exists(filename):
        try:
            with open(filename, 'rb') as f:
                ngan_hang = pickle.load(f)
            print(f"[Đã tải dữ liệu ngân hàng từ '{filename}']")
            return ngan_hang
        except Exception as e:
            print(f"[LỖI] Không thể tải dữ liệu. Bắt đầu với ngân hàng mới. Lỗi: {e}")
            return NganHang()
    else:
        print("[Không tìm thấy file dữ liệu. Bắt đầu với ngân hàng mới.]")
        return NganHang()
def chay_chuong_trinh_cho_tai_khoan(ngan_hang, tai_khoan):
    clear_screen()
    print("\n" + "=" * 45)
    print(f"Chào mừng {tai_khoan.ten_chu_tai_khoan}!")
    print(f"STK: {tai_khoan.so_tai_khoan} | Số dư: {tai_khoan.so_du:,.2f} VND")
    print("=" * 45)
    while True:
        print("\n--- MENU CÁ NHÂN ---")
        print("1. Nạp tiền")
        print("2. Rút tiền")
        print("3. Chuyển tiền")
        print("4. Xem sao kê (trên màn hình)")
        print("5. In sao kê ra file (.txt)")
        print("6. Đăng xuất")
        print("-" * 20)
        choice = nhap_lieu_hop_le("Nhập lựa chọn (1-6)", 1, 6, la_so_nguyen=True)
        if choice is None:
            continue
        try:
            if choice == 1:
                so_tien = nhap_lieu_hop_le("Nhập số tiền muốn nạp (VND)", min_val=0.01)
                if so_tien is not None:
                    thanh_cong, thong_bao = tai_khoan.nap_tien(so_tien)
                    print(thong_bao)
            elif choice == 2:
                so_tien = nhap_lieu_hop_le("Nhập số tiền muốn rút (VND)", min_val=0.01)
                if so_tien is not None:
                    thanh_cong, thong_bao = tai_khoan.rut_tien(so_tien)
                    print(thong_bao)
            elif choice == 3:
                step = 1
                stk_nhan, tk_nhan, so_tien = None, None, None
                while step <= 2:
                    try:
                        if step == 1:
                            print("\n--- Chuyển tiền (Bước 1/2) ---")
                            stk_nhan = nhap_van_ban("Nhập STK người nhận")
                            if stk_nhan is None: break
                            tk_nhan = ngan_hang.tim_tai_khoan(stk_nhan)
                            if tk_nhan is None:
                                print(f"[LỖI] STK {stk_nhan} không tồn tại. Vui lòng thử lại.")
                                continue
                            step += 1
                        elif step == 2:
                            print(f"\n--- Chuyển tiền (Bước 2/2) ---")
                            print(f"Bạn sắp chuyển tiền cho: {tk_nhan.ten_chu_tai_khoan}")
                            so_tien = nhap_lieu_hop_le("Nhập số tiền muốn chuyển (VND)", min_val=0.01)
                            if so_tien is None: break
                            step += 1
                    except LuiMotBuocException:
                        print("\n[LÙI BƯỚC] Quay lại bước trước...")
                        step = max(1, step - 1)
                if stk_nhan and tk_nhan and so_tien:
                    thanh_cong, thong_bao = tai_khoan.chuyen_tien(tk_nhan, so_tien)
                    print(thong_bao)
                else:
                    print("\n[HỦY] Đã hủy thao tác chuyển tiền.")
            elif choice == 4:
                tai_khoan.in_sao_ke()
                clear_screen()
                print("\n" + "=" * 45)
                print(f"Chào mừng {tai_khoan.ten_chu_tai_khoan}!")
                print(f"STK: {tai_khoan.so_tai_khoan} | Số dư: {tai_khoan.so_du:,.2f} VND")
                print("=" * 45)
            elif choice == 5:
                tai_khoan.luu_sao_ke_ra_file()
            elif choice == 6:
                print(f"\nTạm biệt {tai_khoan.ten_chu_tai_khoan}. Hẹn gặp lại!")
                clear_screen()
                break
        except LuiMotBuocException:
            print("\n[LÙI BƯỚC] Không thể lùi thêm. Quay lại menu cá nhân.")
            continue
        except Exception as e:
            print(f"\n[LỖI VẬN HÀNH] Đã xảy ra lỗi: {e}. Vui lòng thử lại.")
def main():
    clear_screen()
    ngan_hang = tai_du_lieu_ngan_hang()
    input("Nhấn Enter để tiếp tục...")
    try:
        while True:
            clear_screen()
            print("\n" + "=" * 40)
            print("  CHÀO MỪNG ĐẾN VỚI NGÂN HÀNG QUOCBANK")
            print("=" * 40)
            print("1. Tạo tài khoản mới")
            print("2. Đăng nhập")
            print("3. Quên mật khẩu")
            print("4. Hiển thị tất cả tài khoản (Admin)")
            print("5. Xem thông tin đăng nhập (Admin)")
            print("6. Xóa toàn bộ dữ liệu (Admin)")
            print("7. Hỗ trợ / Giải đáp lỗi")
            print("8. Thoát chương trình")
            print("-" * 40)
            choice = nhap_lieu_hop_le("Nhập lựa chọn (1-8)", 1, 8, la_so_nguyen=True)
            if choice is None:
                choice = 8
            if choice == 1:
                clear_screen()
                print("\n--- Tạo tài khoản mới ---")
                step = 1
                ten, mat_khau, cau_hoi, tra_loi, so_du_ban_dau = None, None, None, None, None
                tao_tk_thanh_cong = False
                while step <= 5:
                    try:
                        if step == 1:
                            print("\n(Bước 1/5: Tên)")
                            ten = nhap_van_ban("Nhập tên chủ tài khoản", kiem_tra_ten=True)
                            if ten is None: break
                            step += 1
                        elif step == 2:
                            print("\n(Bước 2/5: Mật khẩu)")
                            while True:
                                mat_khau_1 = nhap_van_ban("Nhập mật khẩu (sẽ bị ẩn)", yeu_cau_mat_khau=True)
                                if mat_khau_1 is None: raise StopIteration()
                                mat_khau_2 = nhap_van_ban("Nhập LẠI mật khẩu để xác nhận", yeu_cau_mat_khau=True)
                                if mat_khau_2 is None: raise StopIteration()
                                if mat_khau_1 == mat_khau_2:
                                    mat_khau = mat_khau_1
                                    break
                                else:
                                    print("\n[LỖI] Mật khẩu không khớp. Vui lòng thử lại.")
                            step += 1
                        elif step == 3:
                            print("\n(Bước 3/5: Câu hỏi bảo mật)")
                            cau_hoi = nhap_van_ban("Câu hỏi (VD: Tên con vật cưng đầu tiên?)")
                            if cau_hoi is None: break
                            step += 1
                        elif step == 4:
                            print("\n(Bước 4/5: Câu trả lời)")
                            tra_loi = nhap_van_ban("Câu trả lời (sẽ không bị ẩn)")
                            if tra_loi is None: break
                            step += 1
                        elif step == 5:
                            print("\n(Bước 5/5: Số dư)")
                            so_du_ban_dau = nhap_lieu_hop_le("Nhập số dư ban đầu (VND) (nhập 0 nếu không có)", min_val=0)
                            if so_du_ban_dau is None: break
                            step += 1
                            tao_tk_thanh_cong = True
                    except LuiMotBuocException:
                        print("\n[LÙI BƯỚC] Quay lại bước trước...")
                        step = max(1, step - 1)
                    except StopIteration:
                        break
                if tao_tk_thanh_cong:
                    tk_moi = ngan_hang.them_tai_khoan(ten, mat_khau, so_du_ban_dau, cau_hoi, tra_loi)
                    print("\n✅  TẠO TÀI KHOẢN THÀNH CÔNG!")
                    print(f"  Tên: {tk_moi.ten_chu_tai_khoan}")
                    print(f"  SỐ TÀI KHOẢN: {tk_moi.so_tai_khoan} (Dùng STK này để đăng nhập)")
                    print(f"  Số dư: {tk_moi.so_du:,.2f} VND")
                else:
                    print("\n[HỦY] Đã hủy thao tác tạo tài khoản.")
                input("\nNhấn Enter để quay lại menu chính...")
            elif choice == 2:
                clear_screen()
                print("\n--- Đăng nhập ---")
                try:
                    stk = nhap_van_ban("Nhập Số Tài Khoản")
                    if stk is not None:
                        mat_khau = nhap_van_ban("Nhập Mật khẩu", yeu_cau_mat_khau=True)
                        if mat_khau is not None:
                            tai_khoan_dang_nhap = ngan_hang.dang_nhap(stk, mat_khau)
                            if tai_khoan_dang_nhap:
                                chay_chuong_trinh_cho_tai_khoan(ngan_hang, tai_khoan_dang_nhap)
                            else:
                                print("[LỖI] STK hoặc Mật khẩu không chính xác.")
                                input("\nNhấn Enter để thử lại...")
                except LuiMotBuocException:
                    print("\n[HỦY] Đã hủy đăng nhập.")
            elif choice == 3:
                clear_screen()
                print("\n--- Quên Mật khẩu ---")
                step = 1
                tk, mat_khau_moi = None, None
                khoi_phuc_thanh_cong = False
                while step <= 3:
                    try:
                        if step == 1:
                            print("\n(Bước 1/3: Tìm STK)")
                            stk = nhap_van_ban("Nhập Số Tài Khoản của bạn")
                            if stk is None: break
                            tk = ngan_hang.tim_tai_khoan(stk)
                            if tk:
                                step += 1
                            else:
                                print(f"[LỖI] STK {stk} không tồn tại.")
                        elif step == 2:
                            print("\n(Bước 2/3: Trả lời câu hỏi)")
                            print(f"Câu hỏi bảo mật của bạn là: -> {tk.cau_hoi_bao_mat}")
                            tra_loi = nhap_van_ban("Nhập câu trả lời của bạn")
                            if tra_loi is None: break
                            if tk.kiem_tra_cau_tra_loi(tra_loi):
                                print("\n✅  Xác thực thành công. Vui lòng đặt mật khẩu mới.")
                                step += 1
                            else:
                                print("\n[LỖI] Câu trả lời không chính xác.")
                        elif step == 3:
                            print("\n(Bước 3/3: Đặt mật khẩu mới)")
                            while True:
                                mat_khau_1 = nhap_van_ban("Nhập mật khẩu MỚI (sẽ bị ẩn)", yeu_cau_mat_khau=True)
                                if mat_khau_1 is None: raise StopIteration()
                                mat_khau_2 = nhap_van_ban("Nhập LẠI mật khẩu MỚI để xác nhận", yeu_cau_mat_khau=True)
                                if mat_khau_2 is None: raise StopIteration()
                                if mat_khau_1 == mat_khau_2:
                                    mat_khau_moi = mat_khau_1
                                    break
                                else:
                                    print("\n[LỖI] Mật khẩu không khớp. Vui lòng thử lại.")
                            if mat_khau_moi:
                                tk.mat_khau = mat_khau_moi
                                tk.lich_su.append(f"[{tk._thoi_gian()}] Đặt lại mật khẩu qua câu hỏi bảo mật.")
                                khoi_phuc_thanh_cong = True
                            step += 1
                    except LuiMotBuocException:
                        print("\n[LÙI BƯỚC] Quay lại bước trước...")
                        step = max(1, step - 1)
                    except StopIteration:
                        break
                if khoi_phuc_thanh_cong:
                    print("\n✅  Đã đặt lại mật khẩu thành công.")
                else:
                    print("\n[HỦY] Đã hủy thao tác quên mật khẩu.")
                input("\nNhấn Enter để quay lại menu...")
            elif choice == 4:
                ngan_hang.hien_thi_danh_sach()
            elif choice == 5:
                ngan_hang.hien_thi_thong_tin_dang_nhap()
            elif choice == 6:
                clear_screen()
                print("\n" + "!" * 40)
                print("    CẢNH BÁO NGHIÊM TRỌNG: XÓA DỮ LIỆU")
                print("!" * 40)
                print("\nThao tác này sẽ xóa vĩnh viễn file 'ngan_hang_data.pkl'.")
                print("Toàn bộ tài khoản và lịch sử sẽ bị mất.")
                print("Hệ thống sẽ trở về trạng thái rỗng.")
                try:
                    xac_nhan = nhap_van_ban("\nGõ 'XAC NHAN XOA' để tiếp tục")
                    if xac_nhan == "XAC NHAN XOA":
                        try:
                            data_file = "ngan_hang_data.pkl"
                            if os.path.exists(data_file):
                                os.remove(data_file)
                                print(f"\n✅  Đã xóa file dữ liệu '{data_file}'.")
                            else:
                                print(f"\nKhông tìm thấy file '{data_file}' (đã được xóa từ trước).")
                            ngan_hang = NganHang()
                            print("Đã reset hệ thống ngân hàng về trạng thái rỗng.")
                        except Exception as e:
                            print(f"\n[LỖI] Không thể xóa file. Lý do: {e}")
                    else:
                        print("\nĐã hủy thao tác xóa.")
                except LuiMotBuocException:
                    print("\nĐã hủy thao tác xóa.")
                input("\nNhấn Enter để quay lại menu...")
            elif choice == 7:
                hien_thi_tro_giup()
            elif choice == 8:
                print("\nCảm ơn bạn đã sử dụng QUOCBANK.")
                print("Đang lưu dữ liệu ngân hàng...")
                luu_du_lieu_ngan_hang(ngan_hang)
                break
    except LuiMotBuocException:
        print("\n[LÙI BƯỚC] Không thể lùi thêm. Quay lại menu chính.")
    except KeyboardInterrupt:
        print("\n\nChương trình đã bị hủy bởi người dùng.")
        print("Đang cố gắng lưu dữ liệu trước khi thoát...")
        luu_du_lieu_ngan_hang(ngan_hang)
    except Exception as e:
        print(f"\n[LỖI NGHIÊM TRỌNG] Đã xảy ra lỗi không mong muốn: {e}")
        print("Đang cố gắng lưu dữ liệu khẩn cấp...")
        luu_du_lieu_ngan_hang(ngan_hang)
if __name__ == "__main__":
    main()
```

```python
import datetime
import pickle
import os
import re

def clear_screen():
    os.system('cls' if os.name == 'nt' else 'clear')
def nhap_lieu_hop_le(prompt, min_val=float('-inf'), max_val=float('inf'), la_so_nguyen=False):
    while True:
        try:
            user_input = input(prompt).strip()
            if user_input.lower() in ['exit', 'quit', 'thoat']:
                return None
            if la_so_nguyen:
                value = int(user_input)
            else:
                value = float(user_input)
            if min_val <= value <= max_val:
                return value
            else:
                print(f"Giá trị phải nằm trong khoảng {min_val} đến {max_val}.")
        except ValueError:
            if la_so_nguyen:
                print("Vui lòng nhập một số NGUYÊN hợp lệ.")
            else:
                print("Vui lòng nhập một SỐ hợp lệ.")
        except KeyboardInterrupt:
            print("\nĐã hủy nhập liệu.")
            raise
def nhap_van_ban(prompt):
    while True:
        try:
            user_input = input(prompt).strip()
            if user_input.lower() in ['exit', 'quit', 'thoat']:
                return None
            if not user_input:
                print("Nội dung không được để trống.")
                continue
            return user_input.upper()
        except KeyboardInterrupt:
            print("\nĐã hủy nhập liệu.")
            raise
class KhachHangVay:
    def __init__(self, code, amount, rate, term):
        self.code = code
        self.amount = amount
        self.rate = rate
        self.term = term
        self.interest = self.tinh_lai_don()
        self.total = self.amount + self.interest
    def tinh_lai_don(self):
        return self.amount * self.rate * self.term / 12 / 100
class QuanLyVay:
    def __init__(self):
        self.danh_sach_vay = {}
    def them_hop_dong(self, code, amount, rate, term):
        if code in self.danh_sach_vay:
            print(f"\n[LỖI] Mã HĐ/KH '{code}' đã tồn tại. Không thể thêm.")
            return False
        hop_dong_moi = KhachHangVay(code, amount, rate, term)
        self.danh_sach_vay[code] = hop_dong_moi
        print(f"\nĐã thêm thành công hợp đồng '{code}'.")
        print(f"   - Tiền lãi dự kiến: {hop_dong_moi.interest:,.0f} VND")
        print(f"   - Tổng phải trả: {hop_dong_moi.total:,.0f} VND")
        return True
    def xoa_hop_dong(self, code):
        if code not in self.danh_sach_vay:
            print(f"\n[LỖI] Mã HĐ/KH '{code}' không tồn tại. Không thể xóa.")
            return False
        removed = self.danh_sach_vay.pop(code)
        print(f"\nĐã xóa thành công hợp đồng '{removed.code}' (Gốc: {removed.amount:,.0f} VND).")
        return True
    def tim_hop_dong(self, code):
        return self.danh_sach_vay.get(code)
    def hien_thi_bang_ke(self):
        clear_screen()
        print("\n" + "═" * 95)
        print("║{:^93}║".format("BẢNG KÊ TỔNG HỢP HỢP ĐỒNG VAY"))
        print("╠════════════╦════════════════╦════════════╦════════════╦══════════════════╦══════════════════╣")
        print("║{:^12}║{:^16}║{:^12}║{:^12}║{:^18}║{:^18}║".format(
            "Mã HĐ/KH", "Gốc (VND)", "LS (%/năm)", "Kỳ Hạn (t)", "Lãi (VND)", "Tổng Phải Trả"
        ))
        print("╠════════════╬════════════════╬════════════╬════════════╬══════════════════╬══════════════════╣")
        tong_goc = 0.0
        tong_lai = 0.0
        tong_phai_tra = 0.0
        so_luong_hd = 0
        if not self.danh_sach_vay:
            print("║{:^93}║".format("Chưa có hợp đồng nào trong hệ thống"))
        else:
            for kh in self.danh_sach_vay.values():
                print("║ {:<11}║{:>11,.0f} VND ║{:>9.2f} % ║{:>9} t ║{:>13,.0f} VND ║{:>13,.0f} VND ║".format(
                    kh.code[:11], kh.amount, kh.rate, kh.term, kh.interest, kh.total
                ))
                tong_goc += kh.amount
                tong_lai += kh.interest
                tong_phai_tra += kh.total
                so_luong_hd += 1
        print("╠════════════╩════════════════╩════════════╩════════════╩══════════════════╩══════════════════╣")
        print("║ {:<30} {:>61}║".format(f"Tổng số hợp đồng: {so_luong_hd}", ""))
        print("║ {:<30} {:>57,.0f} VND║".format("Tổng tiền gốc đã cho vay:", tong_goc))
        print("║ {:<30} {:>57,.0f} VND║".format("Tổng lãi dự kiến:", tong_lai))
        print("║ {:<30} {:>57,.0f} VND║".format("Tổng phải thu:", tong_phai_tra))
        print("╚" + "═" * 93 + "╝")
        input("\nNhấn Enter để quay lại menu...")
    def luu_bao_cao_ra_file(self):
        if not self.danh_sach_vay:
            print("\n[LỖI] Không có dữ liệu để xuất báo cáo.")
            return
        filename = f"Bao_Cao_Vay_{datetime.datetime.now().strftime('%Y%m%d_%H%M%S')}.txt"
        try:
            with open(filename, 'w', encoding='utf-8') as f:
                f.write("=" * 78 + "\n")
                f.write(f"{'BÁO CÁO TỔNG HỢP HỢP ĐỒNG VAY':^78}\n")
                f.write(f"Ngày xuất: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
                f.write("=" * 78 + "\n\n")
                f.write(
                    f"{'Mã HĐ/KH':<10} | {'Gốc (VND)':>15} | {'LS (%/năm)':>9} | {'Kỳ Hạn (t)':>9} | {'Lãi (VND)':>15} | {'Tổng Phải Trả':>15}\n")
                f.write("-" * 78 + "\n")
                tong_goc = 0.0
                tong_lai = 0.0
                tong_phai_tra = 0.0
                so_luong_hd = 0
                for kh in self.danh_sach_vay.values():
                    f.write(
                        f"{kh.code:<10} | {kh.amount:>15,.0f} | {kh.rate:>9.2f} | {kh.term:>9} | {kh.interest:>15,.0f} | {kh.total:>15,.0f}\n")
                    tong_goc += kh.amount
                    tong_lai += kh.interest
                    tong_phai_tra += kh.total
                    so_luong_hd += 1
                f.write("=" * 78 + "\n")
                f.write(f"Tổng số hợp đồng: {so_luong_hd}\n")
                f.write(f"Tổng tiền gốc đã cho vay (VND): {tong_goc:,.0f}\n")
                f.write(f"Tổng lãi dự kiến (VND): {tong_lai:,.0f}\n")
                f.write(f"Tổng phải thu (VND): {tong_phai_tra:,.0f}\n")
                f.write("=" * 78 + "\n")
            print(f"\nĐã xuất báo cáo thành công ra file: '{filename}'")
        except Exception as e:
            print(f"\nLỖI: Không thể xuất file. Lý do: {e}")
def luu_du_lieu(quan_ly, filename="quan_ly_vay.pkl"):
    try:
        with open(filename, 'wb') as f:
            pickle.dump(quan_ly, f)
        print(f"\n[Dữ liệu hợp đồng đã được lưu vào '{filename}']")
    except Exception as e:
        print(f"\n[LỖI] Không thể lưu dữ liệu: {e}")
def tai_du_lieu(filename="quan_ly_vay.pkl"):
    if os.path.exists(filename):
        try:
            with open(filename, 'rb') as f:
                quan_ly = pickle.load(f)
            print(f"[Đã tải dữ liệu hợp đồng từ '{filename}']")
            return quan_ly
        except Exception as e:
            print(f"[LỖI] Không thể tải dữ liệu. Bắt đầu với hệ thống mới. Lỗi: {e}")
            return QuanLyVay()
    else:
        print("[Không tìm thấy file dữ liệu. Bắt đầu với hệ thống mới.]")
        return QuanLyVay()
def main():
    clear_screen()
    quan_ly = tai_du_lieu()
    input("Nhấn Enter để tiếp tục...")
    try:
        while True:
            clear_screen()
            print("\n" + "=" * 45)
            print("  HỆ THỐNG QUẢN LÝ HỢP ĐỒNG VAY TÍN CHẤP")
            print("=" * 45)
            print("1. Thêm Hợp đồng Hàng Loạt (n cái)")
            print("2. Thêm Hợp đồng Đơn lẻ")
            print("3. Xóa Hợp đồng Vay")
            print("4. Hiển thị Bảng Kê (Tổng Hợp)")
            print("5. Xuất Báo Cáo ra file (.txt)")
            print("6. Xóa Toàn Bộ Dữ Liệu (Admin)")
            print("7. Thoát")
            print("-" * 45)
            choice = nhap_lieu_hop_le("Nhập lựa chọn (1-7): ", 1, 7, la_so_nguyen=True)
            if choice is None:
                choice = 7
            if choice == 1:
                clear_screen()
                print("\n--- Thêm Hợp đồng Hàng Loạt (n cái) ---")
                print("Gõ 'exit' bất kỳ lúc nào để hủy.")
                n = nhap_lieu_hop_le("Nhập số lượng hợp đồng cần thêm (n): ", min_val=1, la_so_nguyen=True)
                if n is None: continue
                hop_dong_thanh_cong = 0
                hop_dong_that_bai = 0
                for i in range(n):
                    print(f"\n--- Hợp đồng {i + 1} / {n} ---")
                    code = nhap_van_ban("Nhập Mã HĐ/KH: ")
                    if code is None:
                        print("\nĐã hủy nhập hàng loạt.")
                        break
                    if quan_ly.tim_hop_dong(code):
                        print(f"[LỖI] Mã HĐ/KH '{code}' đã tồn tại. Bỏ qua hợp đồng này.")
                        hop_dong_that_bai += 1
                        continue
                    amount = nhap_lieu_hop_le("Nhập số tiền vay (VND): ", min_val=1)
                    if amount is None:
                        print("\nĐã hủy nhập hàng loạt.")
                        break
                    rate = nhap_lieu_hop_le("Nhập lãi suất (%/năm) (ví dụ: 12.5): ", min_val=0.01)
                    if rate is None:
                        print("\nĐã hủy nhập hàng loạt.")
                        break
                    term = nhap_lieu_hop_le("Nhập kỳ hạn (số tháng) (ví dụ: 24): ", min_val=1, la_so_nguyen=True)
                    if term is None:
                        print("\nĐã hủy nhập hàng loạt.")
                        break
                    if quan_ly.them_hop_dong(code, amount, rate, term):
                        hop_dong_thanh_cong += 1
                    else:
                        hop_dong_that_bai += 1
                print("\n" + "=" * 30)
                print("ĐÃ HOÀN TẤT NHẬP HÀNG LOẠT")
                print(f"Thành công: {hop_dong_thanh_cong}")
                print(f"Thất bại/Bỏ qua: {hop_dong_that_bai}")
                input("\nNhấn Enter để quay lại menu...")
            elif choice == 2:
                clear_screen()
                print("\n--- Thêm Hợp đồng Đơn lẻ ---")
                print("Gõ 'exit' bất kỳ lúc nào để hủy.")
                code = nhap_van_ban("Nhập Mã HĐ/KH: ")
                if code is None: continue
                if quan_ly.tim_hop_dong(code):
                    print(f"\n[LỖI] Mã HĐ/KH '{code}' đã tồn tại. Vui lòng nhập mã khác.")
                    input("\nNhấn Enter để quay lại...")
                    continue
                amount = nhap_lieu_hop_le("Nhập số tiền vay (VND): ", min_val=1)
                if amount is None: continue
                rate = nhap_lieu_hop_le("Nhập lãi suất (%/năm): ", min_val=0.01)
                if rate is None: continue
                term = nhap_lieu_hop_le("Nhập kỳ hạn (số tháng): ", min_val=1, la_so_nguyen=True)
                if term is None: continue
                quan_ly.them_hop_dong(code, amount, rate, term)
                input("\nNhấn Enter để quay lại menu...")
            elif choice == 3:
                clear_screen()
                print("\n--- Xóa Hợp đồng Vay ---")
                code = nhap_van_ban("Nhập Mã HĐ/KH cần xóa: ")
                if code is None: continue
                kh = quan_ly.tim_hop_dong(code)
                if kh:
                    print(f"\nBạn sắp xóa hợp đồng sau:")
                    print(f" - Mã: {kh.code}")
                    print(f" - Tiền vay: {kh.amount:,.0f} VND")
                    print(f" - Lãi suất: {kh.rate}%")
                    xac_nhan = nhap_van_ban("Gõ 'y' để xác nhận xóa: ")
                    if xac_nhan.lower() == 'y':
                        quan_ly.xoa_hop_dong(code)
                    else:
                        print("\nĐã hủy thao tác xóa.")
                else:
                    print(f"\n[LỖI] Mã HĐ/KH '{code}' không tồn tại.")
                input("\nNhấn Enter để quay lại menu...")
            elif choice == 4:
                quan_ly.hien_thi_bang_ke()
            elif choice == 5:
                clear_screen()
                print("\n--- Xuất Báo Cáo ra file (.txt) ---")
                quan_ly.luu_bao_cao_ra_file()
                input("\nNhấn Enter để quay lại menu...")
            elif choice == 6:
                clear_screen()
                print("\n" + "!" * 40)
                print("    CẢNH BÁO NGHIÊM TRỌNG: XÓA DỮ LIỆU")
                print("!" * 40)
                print("\nThao tác này sẽ xóa vĩnh viễn file 'quan_ly_vay.pkl'.")
                print("Toàn bộ hợp đồng vay sẽ bị mất.")
                xac_nhan = nhap_van_ban("\nGõ 'XAC NHAN XOA' để tiếp tục (hoặc 'exit' để hủy): ")
                if xac_nhan == "XAC NHAN XOA":
                    try:
                        data_file = "quan_ly_vay.pkl"
                        if os.path.exists(data_file):
                            os.remove(data_file)
                            print(f"\nĐã xóa file dữ liệu '{data_file}'.")
                        else:
                            print(f"\nKhông tìm thấy file '{data_file}' (đã được xóa từ trước).")
                        quan_ly = QuanLyVay()
                        print("Đã reset hệ thống về trạng thái rỗng.")
                    except Exception as e:
                        print(f"\n[LỖI] Không thể xóa file. Lý do: {e}")
                else:
                    print("\nĐã hủy thao tác xóa.")
                input("\nNhấn Enter để quay lại menu...")
            elif choice == 7:
                print("\nCảm ơn bạn đã sử dụng hệ thống.")
                print("Đang lưu dữ liệu...")
                luu_du_lieu(quan_ly)
                break
    except KeyboardInterrupt:
        print("\n\nChương trình đã bị hủy bởi người dùng.")
        print("Đang cố gắng lưu dữ liệu trước khi thoát...")
        luu_du_lieu(quan_ly)
    except Exception as e:
        print(f"\n[LỖI NGHIÊM TRỌNG] Đã xảy ra lỗi không mong muốn: {e}")
        print("Đang cố gắng lưu dữ liệu khẩn cấp...")
        luu_du_lieu(quan_ly)
if __name__ == "__main__":
    main()
```
